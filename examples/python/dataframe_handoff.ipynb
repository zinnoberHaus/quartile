# A dataframe in Quartile
Create a small reproducible dataset, export a typed snapshot, then import it at https://quartile-design.vercel.app/examples/explore. Run from this directory. This is a file handoff; no Python runs in the browser.

```python
import pandas as pd
from quartile_snapshot import write_snapshot

df = pd.DataFrame({
    'sample_id': [f'S-{n:03}' for n in range(60)],
    'batch': ['Control' if n % 2 == 0 else 'Treatment' for n in range(60)],
    'temperature': [None if n % 13 == 0 else 18 + (n * 7 % 30) / 4 for n in range(60)],
    'yield': [42 + (n * 11 % 43) for n in range(60)],
})
df.head()
```

```python
path = write_snapshot(df, 'observations.json', label='Reproducible observations', fields={
    'sample_id': {'type': 'nominal', 'label': 'Sample'},
    'temperature': {'type': 'quantitative', 'label': 'Temperature', 'unit': '°C'},
    'yield': {'type': 'quantitative', 'label': 'Yield'},
})
print(path.resolve())
```
